```python
from docx import Document
```

```python
doc = Document("Dynata Part B 08_08_25.docx")
```

```python
pars_dict = {
    }
i = 1
text_lines = []
for para in doc.paragraphs:
    
    line = para.text.strip()
    line = line.replace("\xa0", " ")


    # if "Scenario" in line:
    #     line = ""

    # for key, value in pars_dict.items():
    #     line = line.replace(key, value)

    line = line.strip()
    
    for subline in line.split("\n"):
        if subline and ((not(subline.startswith("SET"))) and (not (subline.endswith("years"))) and (not (subline.endswith("Scenarios")))):
            if any(char.isdigit() for char in subline):
                text_lines.append("")
                subline = " ".join([f"{i}. ", subline])
                i = i + 1
            text_lines.append(subline)

            if any(char == "?" for char in subline):
                text_lines.append("")

        
for line in text_lines:
    print(line)
```
